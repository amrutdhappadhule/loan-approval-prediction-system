### problem statement

### CIBIL score

````python
CIBIL stands for Credit Information Bureau (India) Limited. 
It is a 3-digit number from 300 to 900 that shows if you pay your debts on time

Score Basics
    Meaning: Shows your financial health and how safe it is for banks to lend you money.
    Good Score: A score of 750 or higher helps you get loans approved fast.
    Low Score: A score under 650 means banks see high risk in lending to you.
````

### What is Credit Risk?

````python
Credit Risk is the possibility that a borrower will fail to repay the loan according to the agreed terms.
    simply:
        What is the chance that this customer wont repay the money?
````

### Factors Banks Consider

````python
1. Income
    Can this customer afford the EMI?
2. Existing EMIs
3. Debt-to-Income Ratio (DTI)
    DTI = Total Monthly Debt Payments / Monthly Income

    Example
        Monthly Income = ₹80,000
        Current EMI = ₹20,000
        New Loan EMI = ₹10,000
        Total Debt = ₹30,000
        DTI  = 30,000 / 80,000
             = 37.5%

4. Credit Score
    A credit score summarises a person's past borrowing behaviour.
5. Repayment History
    Has this customer repaid previous loans on time?
6. Employment Stability
7. Loan Amount
        Suppose
        Income = ₹40,000
        Loan Request = ₹50 lakh
        That request may not align with the applicant's repayment capacity.
        Banks compare the requested loan amount with income and other financial information.
8. Age
    Age also plays a role.
    Examples
    Very young applicants may have limited employment history.
    Applicants close to retirement may have fewer remaining earning years for long loan tenures.
    Banks consider age alongside many other factors
9. Previous Defaults
        
````

````python
first applicant:
    
    | Feature          | Value   |
    | ---------------- | ------- |
    | Income           | ₹90,000 |
    | Credit Score     | 810     |
    | Existing EMI     | ₹15,000 |
    | Previous Default | No      |
    | Job Stability    | 7 years |
    | DTI              | 28%     |

another applicant:
    | Feature          | Value    |
    | ---------------- | -------- |
    | Income           | ₹90,000  |
    | Credit Score     | 560      |
    | Existing EMI     | ₹60,000  |
    | Previous Default | Yes      |
    | Job Stability    | 6 months |
    | DTI              | 67%      |

Despite having the same income,
multiple factors together suggest substantially higher credit risk.
````

### Why can a customer with a high salary still be rejected?

````python
A high salary alone does not guarantee repayment. 
Banks also evaluate credit score, existing debt, repayment history, employment stability, previous defaults, and debt-to-income ratio. 
An applicant with a high income but poor repayment behaviour or excessive existing debt may still represent a high credit risk and 
    therefore may be rejected.
````

### How a Real Bank Makes a Loan Decision

````python
Customer Applies
       │
       ▼
Eligibility Rules
       │
       ▼
Fraud & KYC Checks
       │
       ▼
Credit Risk Assessment
       │
       ▼
Business Rules
       │
       ▼
Human Review (if required)
       │
       ▼
Final Decision
````

### Can a bank rely only on Machine Learning for loan approval?

````python
No. Machine Learning is one component of the underwriting process. 
Banks also apply eligibility rules, KYC verification, fraud detection, regulatory requirements, and business policies. 
High-risk or borderline applications may be reviewed by a human underwriter before a final decision is made.
````

````python
# Step 1: What is a Personal Loan?

A **Personal Loan (PL)** is money borrowed from a bank or NBFC without giving any collateral (like a house or car).

For example,

Suppose Rahul wants ₹5,00,000 for his sister's wedding.

He goes to HDFC Bank.

The bank gives him ₹5,00,000.

Rahul promises to repay it over 5 years with interest.

Since the bank has no collateral, the biggest question is

> **"Will Rahul repay the loan or not?"**

This is exactly the problem banks solve every day.

---

# Step 2: What problem does the bank have?

Imagine 5,000 people apply every day.

The bank cannot manually inspect every application.

They need to answer one question.

> **Should we give this customer the loan?**

If they approve the wrong customer,

the customer may stop paying.

The bank loses money.

If they reject a good customer,

the bank loses business.

So the bank wants the best possible decision.

---

# Step 3: Current Manual Process

Today, many banks still have manual underwriting.

Example

Customer submits

* Aadhaar
* PAN
* Salary slips
* Bank statements
* Employment details

↓

Loan officer verifies

↓

Checks CIBIL

↓

Calculates eligibility

↓

Checks existing loans

↓

Checks income

↓

Makes decision

↓

Approves or rejects

This process can take

* several hours
* one day
* sometimes several days

It also depends on the experience of the loan officer.

Two officers may make different decisions for the same applicant.

That creates inconsistency.

---

# Step 4: What is Underwriting?

One important interview question.

**Underwriting** means

> The process of evaluating the applicant's risk before approving a loan.

The underwriter tries to answer

> "If we lend this money, what is the probability that we will get it back?"

Banks don't lend money.

They lend **risk**.

---

# Step 5: Why Machine Learning?

Suppose the bank has records of

500,000 previous customers.

For every customer they know

Age

Income

Occupation

Loan Amount

Credit Score

EMI

Existing Loans

Loan Status

Default or Not

This historical data becomes training data.

Machine Learning learns

> Which types of customers usually repay successfully.

Then it predicts for new applicants.

---

# Step 6: What exactly is our project?

Our project is **not**

"Predict loan approval."

That is only the final output.

Our real project is

> **Building an AI-powered decision support system for loan underwriting.**

That sounds much closer to how banks describe such systems.

---

# Step 7: What is the real business objective?

The business does **not** care about ML accuracy alone.

The business wants to

* Reduce bad loans (defaults)
* Approve genuine customers faster
* Reduce manual work
* Maintain consistency
* Increase customer satisfaction
* Increase profit
* Reduce operational cost

Machine Learning is only a tool.

The business objective comes first.

---

# Step 8: What are we actually predicting?

Target Variable

Approved

1 = Loan Approved

0 = Loan Rejected

This is a **Binary Classification** problem.

---

# Step 9: What information will the model use?

Think like a loan officer.

If someone asks for ₹10 lakh,

what questions would you ask?

Probably

* Age
* Salary
* Occupation
* Company
* Years of Experience
* Existing EMI
* Existing Loans
* Credit Score
* Previous Defaults
* Monthly Expenses
* Marital Status
* Dependents
* Bank Balance
* Loan Amount
* Loan Purpose
* Employment Type

These become our **features (X)**.

---

# Step 10: What does the model learn?

Suppose historical data shows

Customer A

Income = ₹80,000

Credit Score = 820

No previous defaults

Stable job

Loan Approved

---

Customer B

Income = ₹18,000

Credit Score = 520

Many missed EMIs

Loan Rejected

---

After seeing thousands of examples,

the model starts learning hidden patterns.

It learns that

* Higher income generally lowers risk (though not always).
* Higher credit score is associated with better repayment history.
* Previous defaults increase risk.
* Very high existing EMIs relative to income increase risk.

Notice these are tendencies learned from data, not fixed rules.

---

# Step 11: Why can't we simply approve everyone?

Suppose

100 customers apply.

Each requests ₹10 lakh.

Bank approves everyone.

Total money given

₹10 crore

If

20 customers never repay,

the bank may lose a large amount.

Banks survive by managing risk.

---

# Step 12: Why can't we reject everyone?

No customers

↓

No loans

↓

No interest income

↓

No profit

Banks must balance

**Business Growth**

and

**Risk Management**

---

# Step 13: Where does our project fit?

Think of the workflow.

```
Customer

↓

Loan Application

↓

Upload Documents

↓

Data Verification

↓

Feature Extraction

↓

Machine Learning Model

↓

Risk Score

↓

Approval Recommendation

↓

Loan Officer

↓

Final Decision
```

Notice

The ML model **does not replace** the loan officer.

It supports the decision.

Many banks still keep a human in the loop, especially for higher-value or borderline applications.

---

# Step 14: Is this enough for a real banking project?

Not yet.

A production-grade banking system usually includes much more than a classifier. For example:

* Customer application portal
* Document upload
* OCR for extracting data from documents
* Identity verification (KYC)
* Credit bureau integration (e.g., CIBIL)
* Fraud detection
* Income verification
* Risk scoring engine
* Explainable AI (why the model recommended approval or rejection)
* Decision dashboard for loan officers
* Audit logs
* Model monitoring and retraining

Your ML model is one important component of a larger underwriting system.

---

# Step 15: Our vision for this project

Just as your previous **AI-powered KYC project** aimed to simulate an industry solution rather than a classroom exercise, we'll take the same approach here.

Instead of building only a notebook that predicts **0 or 1**, we'll build an end-to-end **Personal Loan Underwriting & Decision Support System** with:

1. Data validation and preprocessing.
2. Exploratory Data Analysis (EDA).
3. Feature engineering.
4. Model training and evaluation.
5. Risk score generation.
6. Explainable AI (why a decision was made).
7. A decision dashboard.
8. API for predictions.
9. Deployment-ready architecture.
10. Monitoring and logging concepts that mirror real-world ML systems.

This will demonstrate not only machine learning skills but also an understanding of how AI systems are built and used in financial institutions.
````
